```python
from google.colab import userdata

API_KEY = userdata.get("GEMINI_API_KEY")
```

```python
from google import genai

client = genai.Client(api_key=API_KEY)
```

```python
topic = input("Enter Topic: ")
platform = input("Enter Platform: ")
tone = input("Enter Tone: ")
```

```python
prompt = f"""
Generate a social media post.

Topic: {topic}

Platform: {platform}

Tone: {tone}

Platform Rules:

LinkedIn:
- Around 120 words
- Professional
- 3-5 hashtags
- End with a question

Twitter/X:
- Under 280 characters
- Short
- 1-2 hashtags

Instagram:
- Friendly
- Around 80 words
- Emoji
- 5-10 hashtags

Facebook:
- Conversational
- Around 100 words
- 2-4 hashtags
"""
```

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

caption = response.text
print(caption)
```

```python
filename = f"{platform}_{topic.replace(' ','_')}.txt"

with open(filename, "w", encoding="utf-8") as file:
    file.write("Topic: " + topic + "\n")
    file.write("Platform: " + platform + "\n")
    file.write("Tone: " + tone + "\n\n")
    file.write(caption)

print("Saved Successfully")
```
